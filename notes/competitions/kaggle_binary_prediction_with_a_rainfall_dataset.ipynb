```python
import pandas as pd
import seaborn as sns
from catboost import CatBoostClassifier, Pool
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score
```

```python
original_df = pd.read_csv('data/Rainfall.csv')
test_df = pd.read_csv('data/test.csv', index_col='id')
train_df = pd.read_csv('data/train.csv', index_col='id')

target_col = ['rainfall']
x_cols = [col for col in train_df.columns if col not in target_col]
```

```python
X_train, X_test, y_train, y_test = train_test_split(train_df[x_cols], train_df[target_col], test_size=0.2, random_state=20250303)
```

```python
train_pool = Pool(data=X_train, label=y_train,)
test_pool = Pool(data=X_test, label=y_test,)
submit_pool = Pool(data=test_df)
```

```python
base_model = CatBoostClassifier(iterations=5_000)
base_model.fit(train_pool, verbose=500, )
```

```python
preds_class = base_model.predict(test_pool)
```

```python
roc_auc_score(y_test, preds_class)
```
